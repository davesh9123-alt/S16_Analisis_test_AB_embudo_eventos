### 3.1 Estudiar conversión  en las diferentes etapas del embudo

Para estudiar la conversión en las diferentes etapas del embudo (`product_page → product_cart → purchase`), necesitamos analizar cómo los usuarios avanzan por cada etapa. Primero, veamos que eventos tenemos y luego definimos el orden de las etapas del embudo para luego calcular los usuarios unicos en cada etapa del embudo.

Pasos:

- Verificar que eventos tenemos
- Definir el orden de las etapas del embudo
- Calcular los usuarios unicos en cada etapa del embudo
- Calcular las tasas de conversion
- Visualizar graficamente las etapas del embudo

```python
# Primero, veamos qué eventos tenemos en los datos
print("Eventos únicos en el dataset:")
print(events['event_name'].value_counts())

# Definir las etapas del embudo en orden
embudo_etapas = ['login', 'product_page', 'product_cart', 'purchase']
print(f"\nEtapas del embudo a analizar: {embudo_etapas}")
```

```python
# Calcular usuarios únicos que llegaron a cada etapa del embudo
usuarios_por_etapa = {}

for etapa in embudo_etapas:
    usuarios_etapa = events[events['event_name'] == etapa]['user_id'].nunique()
    usuarios_por_etapa[etapa] = usuarios_etapa
    print(f"{etapa}: {usuarios_etapa} usuarios únicos")

# Crear DataFrame para mejor visualización
embudo = pd.DataFrame({
    'etapa': embudo_etapas,
    'usuarios': [usuarios_por_etapa[etapa] for etapa in embudo_etapas]
})

print("\n=== RESUMEN DEL EMBUDO ===")
print(embudo)
```

```python
# Calcular tasas de conversión
embudo['conversion_rate'] = 0.0
embudo['conversion_rate'].iloc[0] = 100.0  # Primera etapa = 100%

for i in range(1, len(embudo)):
    etapa_anterior = embudo['usuarios'].iloc[i-1]
    etapa_actual = embudo['usuarios'].iloc[i]
    conversion = (etapa_actual / etapa_anterior) * 100
    embudo['conversion_rate'].iloc[i] = conversion

# Calcular conversión acumulada (desde el inicio)
total_usuarios_inicio = embudo['usuarios'].iloc[0]
embudo['conversion_acumulada'] = (embudo['usuarios'] / total_usuarios_inicio) * 100

print("\n=== TASAS DE CONVERSIÓN ===")
print(embudo.round(2))
```

```python
# Crear visualización del embudo
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Gráfico 1: Número de usuarios por etapa
ax1.bar(embudo['etapa'], embudo['usuarios'], color=['skyblue', 'lightgreen', 'salmon', 'lightyellow'], edgecolor='black')
ax1.set_title('Usuarios por Etapa del Embudo')
ax1.set_ylabel('Número de Usuarios')
ax1.set_xlabel('Etapa')

# Agregar valores en las barras
for i, v in enumerate(embudo['usuarios']):
    ax1.text(i, v + max(embudo['usuarios'])*0.01, str(v), ha='center', va='bottom')

# Gráfico 2: Tasas de conversión
ax2.plot(embudo['etapa'], embudo['conversion_rate'], embudo['conversion_acumulada'])

ax2.set_title('Conversión por Etapa del Embudo y conversión acumulada')
ax2.set_ylabel('Conversion Rate %')
ax2.set_xlabel('Etapa')
ax2.legend(['conversion_acumulada', 'conversion_rate'])
plt.show()
```

### == Observaciones de las etapas del embudo ==

La conversion muestra numeros favorables. Un 33.34% del total de usuarios en la etapa inicial llegaron a la etapa de compra, lo cual no esta nada mal. De hecho en la ultima etapa la conversion aumento un 1.47% con respecto a la etapa anterior de carrito de compras.

=== TASAS DE CONVERSIÓN ===

- Etapa 1: `login --> usuarios unicos: 58696 --> conversion_rate: 100% --> conversion_acumulada: 100%`
- Etapa 2: `product_page --> usuarios unicos: 38929 --> conversion_rate: 66.32% --> conversion_acumulada: 66.32%`
- Etapa 3: `product_cart --> usuarios unicos: 19284 --> conversion_rate: 49.54% --> conversion_acumulada: 32.85%`
- Etapa 4: `purchase --> usuarios unicos: 19568 --> conversion_rate: 101.47% --> conversion_acumulada: 33.34%`

### 3.2 Estudiar embudo por grupos A y B

```python
# Unir datos de eventos con información de grupos
eventos_con_participantes = events.merge(participants, on='user_id', how='left')

# Separar grupos A y B
eventos_grupo_a = eventos_con_participantes[eventos_con_participantes['group'] == 'A']
eventos_grupo_b = eventos_con_participantes[eventos_con_participantes['group'] == 'B']

# Crear embudos por grupo
embudo_grupo_a = eventos_grupo_a.groupby('event_name')['user_id'].nunique().reset_index().rename(columns={'user_id': 'group_a'})
embudo_grupo_b = eventos_grupo_b.groupby('event_name')['user_id'].nunique().reset_index().rename(columns={'user_id': 'group_b'})

# Unir tablas de embudos por grupo
embudo_en_grupos = embudo_grupo_a.merge(embudo_grupo_b, on='event_name', how='left')

# Mostrar embudo por grupos
embudo_en_grupos = embudo_en_grupos.sort_values('group_b', ascending=False).reset_index().drop(columns='index')
embudo_en_grupos
```

```python

# Graficas separadas
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 8))

# Gráfico 1: Grupo A - Número de usuarios por etapa
ax1.bar(embudo_en_grupos['event_name'], embudo_en_grupos['group_a'], color=['skyblue', 'lightgreen', 'salmon', 'lightyellow'], edgecolor='black')
ax1.set_title('Embudo Grupo A')
ax1.set_ylabel('Número de Usuarios')
ax1.set_xlabel('Etapa')

# Grupo A: Agregar valores en las barras
for i, v in enumerate(embudo_en_grupos['group_a']):
    ax1.text(i, v + max(embudo_en_grupos['group_a'])*0.01, str(v), ha='center', va='bottom')

# Gráfico 2: Grupo A - Número de usuarios por etapa
ax2.bar(embudo_en_grupos['event_name'], embudo_en_grupos['group_b'], color=['skyblue', 'lightgreen', 'salmon', 'lightyellow'], edgecolor='black')
ax2.set_title('Embudo Grupo B')
ax2.set_ylabel('Número de Usuarios')
ax2.set_xlabel('Etapa')

# Grupo B: Agregar valores en las barras
for i, v in enumerate(embudo_en_grupos['group_b']):
    ax2.text(i, v + max(embudo_en_grupos['group_b'])*0.01, str(v), ha='center', va='bottom')

plt.grid(alpha=0.3)
plt.show()
```

### 3.3 Verificar si el número de eventos por usuario está distribuido equitativamente entre las muestras

En este punto debemos verificar si hay equilibrio en la actividad de los usuarios entre los grupos A y B de la prueba A/B. Es crucial verificar esto porque si un grupo tiene usuarios mucho más activos que el otro, podría sesgar los resultados.

Queremos verificar que:

- Los usuarios del grupo A y B tengan niveles similares de actividad
- No haya sesgos donde un grupo tenga usuarios significativamente más activos
- La distribución del número de eventos por usuario sea comparable entre grupos

Pasos para verificar la distribucion de eventos por usuarios entre los grupo A y B:

- Calcular eventos por usuario en cada grupo
- Calcular número de eventos por usuario por grupo
- Comparar distribuciones entre grupos
- Visualizar las distribuciones
- Realizar una prueba estadística para verificar diferencias

```python
# Paso 1: Calcular eventos por usuario en cada grupo

# Calcular número de eventos por usuario por grupo
eventos_por_usuario = eventos_con_participantes.groupby(['user_id', 'group']).size().reset_index(name='num_eventos')

print("=== EVENTOS POR USUARIO POR GRUPO ===")
print(eventos_por_usuario.head(10))

# Estadísticas descriptivas por grupo
stats_por_grupo = eventos_por_usuario.groupby('group')['num_eventos'].describe()
print("\n=== ESTADÍSTICAS DESCRIPTIVAS ===")
print(stats_por_grupo)
```

```python
# Paso 2: Comparar distribuciones entre grupos

# Separar datos por grupo
grupo_A = eventos_por_usuario[eventos_por_usuario['group'] == 'A']['num_eventos']
grupo_B = eventos_por_usuario[eventos_por_usuario['group'] == 'B']['num_eventos']

print(f"\n=== COMPARACIÓN DE GRUPOS ===")
print(f"Grupo A - Media: {grupo_A.mean():.2f}, Mediana: {grupo_A.median():.2f}")
print(f"Grupo B - Media: {grupo_B.mean():.2f}, Mediana: {grupo_B.median():.2f}")
print(f"Diferencia en medias: {abs(grupo_A.mean() - grupo_B.mean()):.2f}")
```

```python
# Paso 4: Prueba estadística para verificar diferencias

# Realizar prueba t de Student para comparar medias

# Prueba t para muestras independientes
t_stat, p_value = st.ttest_ind(grupo_A, grupo_B)

print(f"\n=== PRUEBA ESTADÍSTICA ===")
print(f"Estadístico t: {t_stat:.4f}")
print(f"Valor p: {p_value:.4f}")

# Interpretar resultado
alpha = 0.05
if p_value < alpha:
    print(f"❌ HAY diferencia significativa entre grupos (p < {alpha})")
    print("   Los grupos NO están equilibrados en actividad")
else:
    print(f"✅ NO hay diferencia significativa entre grupos (p >= {alpha})")
    print("   Los grupos ESTÁN equilibrados en actividad")
```

### == Observaciones de la prueba preliminar ==

Las medias del grupo A y B son relativamente similares (`medias: A=7.46 y B=7.11`). Finalmente la prueba nos arroga un resultado algo inesperado. El valor P es significativamente inferior a `alpha=0.05` lo que quiere decir que los grupos no estan equilibrados. Debemos seguir explorando los datos para encontrar el motivo de desequilibrio.

```python
# Crear tabla resumen más detallada
resumen_distribucion = pd.DataFrame({
    'Grupo A': [
        len(grupo_A),
        grupo_A.mean(),
        grupo_A.median(),
        grupo_A.std(),
        grupo_A.min(),
        grupo_A.max()
    ],
    'Grupo B': [
        len(grupo_B),
        grupo_B.mean(),
        grupo_B.median(),
        grupo_B.std(),
        grupo_B.min(),
        grupo_B.max()
    ]
}, index=['Usuarios', 'Media', 'Mediana', 'Desv. Estándar', 'Mínimo', 'Máximo'])

print("\n=== RESUMEN DETALLADO ===")
print(resumen_distribucion.round(2))

# Calcular diferencia porcentual entre medias
diff_porcentual = abs(grupo_A.mean() - grupo_B.mean()) / grupo_A.mean() * 100
print(f"\nDiferencia porcentual entre medias: {diff_porcentual:.2f}%")

if diff_porcentual < 5:
    print("✅ Diferencia menor al 5% - Distribución equilibrada")
elif diff_porcentual < 10:
    print("⚠️ Diferencia entre 5-10% - Revisar posibles sesgos")
else:
    print("❌ Diferencia mayor al 10% - Distribución desequilibrada")
```

### == Observaciones ==

```python
# Analizar distribución por percentiles
percentiles = [25, 50, 75, 90, 95]
print(f"\n=== ANÁLISIS DE PERCENTILES ===")
print("Percentil\tGrupo A\t\tGrupo B\t\tDiferencia")
print("-" * 50)

for p in percentiles:
    perc_A = np.percentile(grupo_A, p)
    perc_B = np.percentile(grupo_B, p)
    diff = abs(perc_A - perc_B)
    print(f"P{p}\t\t{perc_A:.2f}\t\t{perc_B:.2f}\t\t{diff:.2f}")

# Crear rangos de actividad y analizar distribución
def categorizar_actividad(num_eventos):
    if num_eventos <= 5:
        return 'Baja'
    elif num_eventos <= 15:
        return 'Media'
    else:
        return 'Alta'

# Aplicar categorización
eventos_por_usuario['categoria_actividad'] = eventos_por_usuario['num_eventos'].apply(categorizar_actividad)

# Analizar distribución por categorías
distribucion_categorias = eventos_por_usuario.groupby(['group', 'categoria_actividad']).size().unstack(fill_value=0)
distribucion_porcentajes = distribucion_categorias.div(distribucion_categorias.sum(axis=1), axis=0) * 100

print(f"\n=== DISTRIBUCIÓN POR CATEGORÍAS DE ACTIVIDAD ===")
print("Conteos absolutos:")
print(distribucion_categorias)
print("\nPorcentajes:")
print(distribucion_porcentajes.round(2))
```

### == Observaciones de las graficas ==

Las graficas de la distribucion de categorias de actividad muestran un similitud muy marcada, pero como ya comprobamos con la prueba estadistica esta similitud es solo aparente. Podemos probar hacer una prueba adicional para comprobar las diferencias.

```python
# Prueba adicional para comparar distribuciones completas
from scipy.stats import ks_2samp

# Prueba de Kolmogorov-Smirnov
ks_stat, ks_p_value = ks_2samp(grupo_A, grupo_B)

print(f"\n=== PRUEBA DE KOLMOGOROV-SMIRNOV ===")
print(f"Estadístico KS: {ks_stat:.4f}")
print(f"Valor p: {ks_p_value:.4f}")

if ks_p_value < 0.05:
    print("❌ Las distribuciones son significativamente diferentes")
else:
    print("✅ Las distribuciones son similares")
```

```python
print("\n" + "="*60)
print("                    CONCLUSIONES FINALES")
print("="*60)

# Resumen de todas las métricas
print(f"📊 MÉTRICAS CLAVE:")
print(f"   • Diferencia en medias: {abs(grupo_A.mean() - grupo_B.mean()):.2f} eventos")
print(f"   • Diferencia porcentual: {diff_porcentual:.2f}%")
print(f"   • Prueba t (p-value): {p_value:.4f}")
print(f"   • Prueba KS (p-value): {ks_p_value:.4f}")

print(f"\n🎯 EVALUACIÓN DE EQUILIBRIO:")

# Criterios de evaluación
criterios_cumplidos = 0
total_criterios = 4

if diff_porcentual < 10:
    print("   ✅ Diferencia porcentual aceptable (< 10%)")
    criterios_cumplidos += 1
else:
    print("   ❌ Diferencia porcentual alta (>= 10%)")

if p_value >= 0.05:
    print("   ✅ No hay diferencia significativa en medias")
    criterios_cumplidos += 1
else:
    print("   ❌ Diferencia significativa en medias")

if ks_p_value >= 0.05:
    print("   ✅ Distribuciones similares (KS test)")
    criterios_cumplidos += 1
else:
    print("   ❌ Distribuciones diferentes (KS test)")

# Evaluar equilibrio en categorías de actividad
max_diff_categoria = 0
for categoria in distribucion_porcentajes.columns:
    diff_cat = abs(distribucion_porcentajes.loc['A', categoria] - distribucion_porcentajes.loc['B', categoria])
    max_diff_categoria = max(max_diff_categoria, diff_cat)

if max_diff_categoria < 5:
    print("   ✅ Distribución equilibrada por categorías de actividad")
    criterios_cumplidos += 1
else:
    print("   ❌ Desequilibrio en categorías de actividad")

# Evaluación final
porcentaje_cumplimiento = (criterios_cumplidos / total_criterios) * 100

print(f"\n📈 PUNTUACIÓN DE EQUILIBRIO: {criterios_cumplidos}/{total_criterios} ({porcentaje_cumplimiento:.1f}%)")

if porcentaje_cumplimiento >= 75:
    conclusion = "✅ DISTRIBUCIÓN EQUILIBRADA"
    recomendacion = "Los grupos están bien balanceados. Puedes proceder con confianza al análisis de la prueba A/B."
elif porcentaje_cumplimiento >= 50:
    conclusion = "⚠️ DISTRIBUCIÓN PARCIALMENTE EQUILIBRADA"
    recomendacion = "Hay algunos desequilibrios menores. Considera esto al interpretar los resultados."
else:
    conclusion = "❌ DISTRIBUCIÓN DESEQUILIBRADA"
    recomendacion = "Los grupos muestran diferencias significativas. Revisa la aleatorización o considera estratificación."

print(f"\n🏆 CONCLUSIÓN: {conclusion}")
print(f"\n💡 RECOMENDACIÓN:")
print(f"   {recomendacion}")
```

### 4.5 Aplicar prueba z para cada etapa del embudo

En esta parte queremos comparar las proporciones para cada etapa del embudo entre los grupos A y B. Lo que queremos comprobar es si realmente hubo esa mejoria esperada del 10%, que hasta ahora parece que no existe ese incremento. Debemos definir un hipotesis nula y una hipotesis alternativa.

- H₀: No hay diferencia en las proporciones entre grupos
- H₁: Hay diferencia significativa entre grupos

Pasos:

- 1. Crear una funcion para la prueba z de proporciones
- 2. Aplicar la prueba para cada grupo y para cada etapa del embudo
- 3. Verificar resultados

```python
# Paso 1:
from scipy.stats import norm

def prueba_z_proporciones(x1, n1, x2, n2):
    """
    Prueba z para diferencia de proporciones
    x1, x2: número de éxitos en cada grupo
    n1, n2: tamaño de muestra de cada grupo
    """
    # Calcular proporciones
    p1 = x1 / n1
    p2 = x2 / n2
    
    # Proporción combinada
    p_combined = (x1 + x2) / (n1 + n2)
    
    # Error estándar
    se = np.sqrt(p_combined * (1 - p_combined) * (1/n1 + 1/n2))
    
    # Estadístico z
    z_stat = (p1 - p2) / se
    
    # P-value (prueba de dos colas)
    p_value = 2 * (1 - norm.cdf(abs(z_stat)))
    
    return z_stat, p_value, p1, p2
```

```python
# Paso 2: Aplicar la prueba para cada grupo y para cada etapa del embudo

# Para cada evento en el embudo
eventos_embudo = ['login', 'product_page', 'product_cart', 'purchase']

# Aplicar prueba z para cada evento
resultados_pruebas = []

for i, evento in enumerate(eventos_embudo):
    # Definir éxitos y totales para cada grupo usando índices numéricos
    
    # x1 = usuarios grupo A que llegaron al evento
    x1 = embudo_en_grupos_clean['group_a'][i]
    # n1 = total usuarios grupo A (siempre el primer valor - login)
    n1 = embudo_en_grupos_clean['group_a'][0]
    
    # x2 = usuarios grupo B que llegaron al evento  
    x2 = embudo_en_grupos_clean['group_b'][i]
    # n2 = total usuarios grupo B (siempre el primer valor - login)
    n2 = embudo_en_grupos_clean['group_b'][0]
    
    # Aplicar prueba z
    z_stat, p_value, p1, p2 = prueba_z_proporciones(x1, n1, x2, n2)
    
    # Guardar resultados
    resultados_pruebas.append({
        'evento': evento,
        'grupo_a_usuarios': x1,
        'grupo_b_usuarios': x2,
        'proporcion_a': p1,
        'proporcion_b': p2,
        'diferencia': p1 - p2,
        'z_stat': z_stat,
        'p_value': p_value,
        'significativo': p_value < 0.05
    })
    
    # Mostrar resultados para cada evento
    print(f"\n=== {evento.upper()} ===")
    print(f"Grupo A: {x1}/{n1} = {p1:.4f} ({p1*100:.2f}%)")
    print(f"Grupo B: {x2}/{n2} = {p2:.4f} ({p2*100:.2f}%)")
    print(f"Diferencia: {(p1-p2)*100:.2f} puntos porcentuales")
    print(f"Z-statistic: {z_stat:.4f}")
    print(f"P-value: {p_value:.4f}")
    print(f"¿Significativo? {'Sí' if p_value < 0.05 else 'No'}")

# Crear DataFrame con todos los resultados
df_resultados = pd.DataFrame(resultados_pruebas)
print(f"\n=== RESUMEN DE TODAS LAS PRUEBAS ===")
print(df_resultados.round(4))
```
